## 1. Python Syntax

Python uses indentation to define blocks of code.

```python
if condition:
    # indented code
    print("Hello")
```

### Important rules
- Python is case-sensitive.
- Indentation matters.
- Usually use **4 spaces** for indentation.
- Comments start with `#`.
- Statements normally do not need `;`.

Example:

````python
# This is a comment
print("Hello, Python!")

if True:
    print("Indentation is important")
````

## 2. Variables

A variable stores a value.

```python
name = "Divya"
age = 20
```

Python automatically determines the type of the value.

### Naming rules
- Can contain letters, numbers, and `_`
- Cannot start with a number
- Cannot use Python keywords
- Use meaningful names

Good:
```python
student_name = "Divya"
total_marks = 95
```

Avoid:
```python

a1 = 95
```

when a more descriptive name is possible.

````python
name = "Divya"
age = 20
course = "Computer Engineering"

print(name)
print(age)
print(course)
````

## 3. Data Types

Common Python data types:

| Type | Example | Meaning |
|---|---|---|
| `int` | `10` | Integer |
| `float` | `10.5` | Decimal number |
| `str` | `"Python"` | Text |
| `bool` | `True` | True/False |
| `list` | `[1, 2, 3]` | Ordered, mutable collection |
| `tuple` | `(1, 2, 3)` | Ordered, immutable collection |
| `set` | `{1, 2, 3}` | Unique values |
| `dict` | `{"name": "Divya"}` | Key-value pairs |
| `None` | `None` | No value |

Use `type()` to check a value's type.

````python
integer_value = 10
float_value = 10.5
text_value = "Python"
boolean_value = True
list_value = [1, 2, 3]
tuple_value = (1, 2, 3)
set_value = {1, 2, 3}
dict_value = {"name": "Divya"}

print(type(integer_value))
print(type(float_value))
print(type(text_value))
print(type(boolean_value))
print(type(list_value))
print(type(tuple_value))
print(type(set_value))
print(type(dict_value))
````

## 7. Strings

A string is text enclosed in quotes.

```python
name = "Python"
```

### Useful operations
- Indexing
- Slicing
- Concatenation
- Formatting
- String methods

Python strings are **immutable**: you cannot directly change one character of an existing string.

````python
text = "Python Programming"

print(text[0])       # First character
print(text[-1])      # Last character
print(text[0:6])     # Slicing
print(text.upper())
print(text.lower())
print(text.replace("Python", "AI"))
print(len(text))
````

### f-Strings


## 8. Lists

A list stores multiple values in an ordered, mutable collection.

```python
fruits = ["apple", "banana", "mango"]
```

Common methods:
- `append()`
- `insert()`
- `remove()`
- `pop()`
- `sort()`
- `reverse()`
- `extend()`

Lists support indexing and slicing.

````python
fruits = ["apple", "banana", "mango"]

print(fruits[0])
print(fruits[-1])

fruits.append("orange")
fruits.insert(1, "grapes")
fruits.remove("banana")

print(fruits)

last_item = fruits.pop()
print("Removed:", last_item)
print("Final list:", fruits)
````

## 9. Tuples

A tuple is ordered but **immutable**.

Use tuples when the collection should not be changed.

```python
coordinates = (19.0760, 72.8777)
```

````python
coordinates = (19.0760, 72.8777)

print(coordinates[0])
print(coordinates[1])

# coordinates[0] = 20  # This would raise an error
````

## 10. Sets

A set stores **unique** values and does not preserve duplicates.

Useful for removing duplicates and membership checks.

````python
numbers = {1, 2, 2, 3, 4, 4}
print(numbers)

numbers.add(5)
numbers.remove(1)

print(numbers)
````

## 11. Dictionaries

A dictionary stores data as **key-value pairs**.

```python
student = {
    "name": "Divya",
    "age": 20
}
```

Common methods:
- `keys()`
- `values()`
- `items()`
- `get()`
- `update()`
- `pop()`

````python
student = {
    "name": "Divya",
    "age": 20,
    "branch": "Computer Engineering"
}

print(student["name"])
print(student.get("age"))

student["age"] = 21
student["city"] = "Mumbai"

print(student.keys())
print(student.values())
print(student.items())
````

## 12. Conditional Statements

Use `if`, `elif`, and `else` to make decisions.

```python
if condition:
    ...
elif another_condition:
    ...
else:
    ...
```

````python
marks = 78

if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
else:
    grade = "C"

print("Grade:", grade)
````

## 13. For Loops

A `for` loop is commonly used to iterate over a sequence.

```python
for item in collection:
    print(item)
```

`range()` generates a sequence of numbers.

```python
range(start, stop, step)
```

The `stop` value is not included.

````python
for number in range(1, 6):
    print(number)

print("Fruits:")

fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
````

## 14. While Loops

A `while` loop continues as long as its condition is `True`.

Be careful to update the condition so the loop does not become infinite.

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

## 15. break, continue, pass

- `break` stops the loop.
- `continue` skips the current iteration.
- `pass` does nothing; it is useful as a placeholder.

````python
for number in range(1, 11):
    if number == 4:
        continue

    if number == 8:
        break

    print(number)

# pass example
if True:
    pass
````

## 16. Functions

A function is a reusable block of code.

Syntax:
```python
def function_name(parameters):
    # code
    return value
```

### Why functions?
- Reuse code
- Keep programs organized
- Make debugging easier
- Improve readability

````python
def greet(name):
    return f"Hello, {name}!"

message = greet("Divya")
print(message)
````

### Function Parameters

Python supports:
- Positional arguments
- Keyword arguments
- Default arguments

Example:

````python
def introduce(name, course="Computer Engineering"):
    return f"{name} studies {course}."

print(introduce("Divya"))
print(introduce("Divya", "Artificial Intelligence"))
print(introduce(name="Divya", course="Python"))
````

## 17. Scope

### Local variable
Created inside a function and normally available only inside it.

### Global variable
Created outside functions.

Prefer passing values into functions and returning results rather than relying heavily on global variables.

````python
course = "Python"

def show_course():
    local_message = "Learning Python"
    print(course)
    print(local_message)

show_course()

# local_message is not available here
````

## 18. List Comprehension

List comprehension provides a concise way to create lists.

Normal:
```python
squares = []
for n in range(5):
    squares.append(n * n)
```

Comprehension:
```python
squares = [n * n for n in range(5)]
```

````python
squares = [n * n for n in range(1, 6)]
even_numbers = [n for n in range(1, 11) if n % 2 == 0]

print(squares)
print(even_numbers)
````

## 19. Exception Handling

Errors can occur while a program runs.

Use:
- `try`
- `except`
- `else`
- `finally`

Do not hide errors unnecessarily. Handle expected errors clearly.

````python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Number cannot be zero.")

else:
    print("Result:", result)

finally:
    print("Execution completed.")
````

## 20. File Handling

Use `open()` to work with files.

Common modes:
- `"r"` — read
- `"w"` — write/overwrite
- `"a"` — append
- `"x"` — create

Prefer `with open(...)` because it automatically closes the file.

````python
# Write a file
with open("example.txt", "w") as file:
    file.write("Hello from Python!")

# Read the file
with open("example.txt", "r") as file:
    content = file.read()

print(content)
````

## 24. `enumerate()`

Use `enumerate()` when you need both the index and value while looping.

````python
fruits = ["apple", "banana", "mango"]

for index, fruit in enumerate(fruits, start=1):
    print(index, fruit)
````

## 25. `zip()`

`zip()` combines items from multiple iterables.

It is useful when two lists contain related information.

````python
names = ["A", "B", "C"]
scores = [90, 85, 95]

for name, score in zip(names, scores):
    print(name, score)
````

## 26. Mutable vs Immutable

### Mutable
Can be changed after creation:
- `list`
- `dict`
- `set`

### Immutable
Cannot be changed after creation:
- `int`
- `float`
- `str`
- `bool`
- `tuple`

This distinction is important when passing objects to functions and working with data.

## 27. `==` vs `is`

`==` checks whether two values are equal.

`is` checks whether two references point to the same object.

For normal value comparisons, use `==`.

Use `is` commonly for identity checks such as:

```python
if value is None:
    ...
```

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)  # Same contents
print(a is b)  # Different list objects
````

## 28. Object-Oriented Programming — Basic Idea

Python supports OOP.

A **class** is a blueprint and an **object** is an instance of that class.

You will use classes frequently in real projects and libraries.

````python
class Student:
    def __init__(self, name, branch):
        self.name = name
        self.branch = branch

    def introduce(self):
        return f"I am {self.name} from {self.branch}."

student = Student("Divya", "Computer Engineering")

print(student.introduce())
````
