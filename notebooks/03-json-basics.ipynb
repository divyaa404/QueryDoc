## 1. What is JSON?

**JSON = JavaScript Object Notation**

JSON is a lightweight text format used to store and exchange data.

Although its name contains "JavaScript", JSON is language-independent and is widely used with:

- Python
- JavaScript
- Java
- C++
- Go
- PHP
- Flutter/Dart
- AI applications
- REST APIs

Example:

```json
{
  "name": "Divya",
  "age": 20,
  "branch": "Computer Engineering"
}
```

The data is easy for both humans and programs to read.

## 2. Why is JSON Important?

JSON is extremely common in modern software development.

For example:

```text
Python Application
```

You will encounter JSON when working with:

## 3. Basic JSON Structure

A JSON object uses curly braces:

```json
{
  "name": "Divya",
  "age": 20
}
```

Each entry has:

```text
"key": value
```

Multiple entries are separated by commas.

### Important

Keys in standard JSON objects must be written using **double quotes**.

Correct:

```json
{"name": "Divya"}
```

Not valid JSON:

```text
{'name': 'Divya'}
```

The second example resembles a Python dictionary, but it is not standard JSON.

## 4. JSON Data Types

JSON supports these basic value types:

| JSON Type | Example |
|---|---|
| String | `"Python"` |
| Number | `20` |
| Boolean | `true` |
| Null | `null` |
| Object | `{"name": "Divya"}` |
| Array | `["Python", "AI"]` |

Notice that JSON uses:

```text
true
false
null
```

while Python uses:

```text
True
False
None
```

## 5. JSON Strings

JSON strings use double quotes.

Example:

```json
{
  "name": "Divya",
  "course": "Computer Engineering"
}
```

A string can contain spaces and many normal characters.

JSON strings can also contain escaped characters.

Example:

```json
{
  "message": "Hello\nPython"
}
```

## 6. JSON Numbers

JSON supports numbers such as:

```json
{
  "age": 20,
  "price": 99.50,
  "score": 95
}
```

JSON does not distinguish between Python's `int` and `float` in exactly the same way; when JSON is parsed by Python, numbers become appropriate Python numeric types.

## 7. JSON Boolean

JSON uses lowercase:

```json
{
  "is_student": true,
  "is_admin": false
}
```

Python uses:

```python
True
False
```

This difference is important when manually writing JSON.

## 8. JSON Null

JSON uses:

```json
null
```

to represent an empty or missing value.

Example:

```json
{
  "name": "Divya",
  "middle_name": null
}
```

When parsed into Python, JSON `null` becomes Python `None`.

## 9. JSON Arrays

A JSON array uses square brackets:

```json
{
  "skills": [
    "Python",
    "Flutter",
    "AI"
  ]
}
```

Arrays are similar to Python lists.

They can contain multiple values:

```json
[10, 20, 30]
```

They can also contain objects:

```json
[
  {"name": "A", "age": 20},
  {"name": "B", "age": 21}
]
```

## 10. Nested JSON

JSON objects can contain other objects and arrays.

Example:

```json
{
  "name": "Divya",
  "education": {
    "college": "SLRTCE",
    "branch": "Computer Engineering"
  },
  "skills": [
    "Python",
    "Flutter",
    "AI"
  ]
}
```

This is called **nested JSON**.

Real APIs often return deeply nested JSON, so learning to navigate it is important.

````python
student = {
    "name": "Divya",
    "education": {
        "college": "SLRTCE",
        "branch": "Computer Engineering"
    },
    "skills": [
        "Python",
        "Flutter",
        "AI"
    ]
}

print(student["name"])
print(student["education"]["college"])
print(student["education"]["branch"])
print(student["skills"][0])
````

## 11. JSON vs Python Dictionary

They look similar but are not exactly the same.

### Python dictionary

```python
student = {
    "name": "Divya",
    "age": 20,
    "is_student": True,
    "middle_name": None
}
```

### JSON

```json
{
  "name": "Divya",
  "age": 20,
  "is_student": true,
  "middle_name": null
}
```

Main differences:

| Python | JSON |
|---|---|
| `True` | `true` |
| `False` | `false` |
| `None` | `null` |
| Often single or double quoted strings | Double-quoted strings |
| Dictionary | Object |
| List | Array |

JSON is text. A Python dictionary is a Python object.

## 13. `json.dumps()`

`json.dumps()` converts a Python object into a JSON-formatted string.

Think:

```text
Python → JSON string
```

Example:

````python
import json

student = {
    "name": "Divya",
    "age": 20,
    "skills": ["Python", "AI"]
}

json_string = json.dumps(student)

print(json_string)
print(type(json_string))
````

## 14. Pretty JSON

You can make JSON easier to read using `indent`.

```python
json.dumps(data, indent=4)
```

````python
import json

student = {
    "name": "Divya",
    "age": 20,
    "skills": ["Python", "AI"]
}

pretty_json = json.dumps(student, indent=4)

print(pretty_json)
````

## 15. `json.loads()`

`json.loads()` converts a JSON string into a Python object.

Think:

```text
JSON string → Python object
```

Example:

````python
import json

json_string = '{"name": "Divya", "age": 20}'

student = json.loads(json_string)

print(student)
print(type(student))
print(student["name"])
print(student["age"])
````

## 16. `loads()` vs `dumps()`

Remember this:

```text
loads  → JSON string → Python
dumps  → Python → JSON string
```

A useful memory trick:

- **load** JSON into Python → `loads()`
- **dump** Python into JSON → `dumps()`

The `s` means the operation is working with a **string**.

## 17. `json.load()` and `json.dump()`

There are also versions without `s`:

```text
load  → read JSON from a file
dump  → write JSON to a file
```

So:

| Function | Purpose |
|---|---|
| `json.loads()` | JSON string → Python |
| `json.dumps()` | Python → JSON string |
| `json.load()` | JSON file → Python |
| `json.dump()` | Python → JSON file |

## 18. Writing a JSON File

Use `json.dump()` to write Python data to a `.json` file.

````python
import json

student = {
    "name": "Divya",
    "age": 20,
    "branch": "Computer Engineering",
    "skills": ["Python", "Flutter", "AI"]
}

with open("student.json", "w", encoding="utf-8") as file:
    json.dump(student, file, indent=4)

print("student.json created")
````

## 19. Reading a JSON File

Use `json.load()` to read a JSON file into Python.

````python
import json

with open("student.json", "r", encoding="utf-8") as file:
    student = json.load(file)

print(student)
print("Name:", student["name"])
print("Skills:", student["skills"])
````

## 20. Accessing JSON Data

Suppose an API returns:

```json
{
  "id": 1,
  "name": "Divya",
  "profile": {
    "city": "Mumbai",
    "skills": ["Python", "AI"]
  }
}
```

In Python:

```python
data["id"]
data["name"]
data["profile"]["city"]
data["profile"]["skills"][0]
```

For nested JSON, move through each level step by step.

````python
data = {
    "id": 1,
    "name": "Divya",
    "profile": {
        "city": "Mumbai",
        "skills": ["Python", "AI"]
    }
}

print(data["id"])
print(data["name"])
print(data["profile"]["city"])
print(data["profile"]["skills"][0])
````

## 21. The `get()` Method

Direct access:

```python
data["name"]
```

can raise a `KeyError` if the key does not exist.

Using:

```python
data.get("name")
```

returns `None` by default if the key is missing.

You can also provide a default value:

```python
data.get("city", "Unknown")
```

````python
data = {
    "name": "Divya",
    "age": 20
}

print(data.get("name"))
print(data.get("city"))
print(data.get("city", "Unknown"))
````

## 22. Updating JSON-like Python Data

Once JSON has been converted to Python, you can modify the Python object.

Example:

````python
student = {
    "name": "Divya",
    "age": 20,
    "skills": ["Python"]
}

student["age"] = 21
student["skills"].append("AI")
student["city"] = "Mumbai"

print(student)
````

## 23. Removing Data

Python dictionaries provide methods such as:

```python
pop()
del
```

Example:

````python
student = {
    "name": "Divya",
    "age": 20,
    "city": "Mumbai"
}

removed_city = student.pop("city")

print("Removed:", removed_city)
print(student)
````

## 24. Working with JSON Arrays

When an API returns a JSON array, Python usually receives a list.

Example JSON:

```json
[
  {"id": 1, "name": "A"},
  {"id": 2, "name": "B"},
  {"id": 3, "name": "C"}
]
```

You can loop through it:

````python
users = [
    {"id": 1, "name": "A"},
    {"id": 2, "name": "B"},
    {"id": 3, "name": "C"}
]

for user in users:
    print(user["id"], user["name"])
````

## 25. JSON from an API

In the previous API notebook, we used:

```python
response = requests.get(url)
```

If the API returns JSON, you can use:

```python
data = response.json()
```

This is different from manually using `json.loads(response.text)` because `response.json()` is the convenient interface provided by `requests` for JSON responses.

````python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.get(url, timeout=10)
response.raise_for_status()

data = response.json()

print(type(data))
print(data)
print("Title:", data["title"])
````

## 27. JSON Error Handling

Invalid JSON can raise:

```python
json.JSONDecodeError
```

Example:

````python
import json

invalid_json = '{"name": "Divya", "age": 20'

try:
    data = json.loads(invalid_json)
    print(data)
except json.JSONDecodeError as error:
    print("Invalid JSON:", error)
````

## 28. Common JSON Mistakes

### Mistake 1 — Single quotes

Not standard JSON:

```text
{'name': 'Divya'}
```

Correct:

```json
{"name": "Divya"}
```

### Mistake 2 — Missing comma

Incorrect:

```json
{
  "name": "Divya"
  "age": 20
}
```

Correct:

```json
{
  "name": "Divya",
  "age": 20
}
```

### Mistake 3 — Python boolean

Incorrect JSON:

```text
{"active": True}
```

Correct JSON:

```json
{"active": true}
```

### Mistake 4 — Python None

Incorrect JSON:

```text
{"value": None}
```

Correct JSON:

```json
{"value": null}
```

### Mistake 5 — Trailing comma

Be careful with trailing commas when writing strict JSON.

## 29. JSON Schema — Basic Idea

As applications grow, developers often need to define what a JSON document should look like.

For example, a user object might be expected to contain:

```text
id       → number
name     → string
email    → string
age      → number
```

**JSON Schema** is a standard for describing and validating JSON structures.

You do not need to learn JSON Schema deeply yet. Just understand that APIs can define expected JSON structures.

# Quick JSON Cheat Sheet

```python
import json

# Python dictionary
data = {
    "name": "Divya",
    "age": 20
}

# Python → JSON string
json_string = json.dumps(data)

# Pretty JSON
json_string = json.dumps(data, indent=4)

# JSON string → Python
data = json.loads(json_string)

# Write JSON file
with open("data.json", "w", encoding="utf-8") as file:
    json.dump(data, file, indent=4)

# Read JSON file
with open("data.json", "r", encoding="utf-8") as file:
    data = json.load(file)

# Access data
print(data["name"])

# Safe access
print(data.get("name"))
print(data.get("city", "Unknown"))

# API JSON
response = requests.get(url, timeout=10)
response.raise_for_status()
data = response.json()
```
