## 1. What is an API?

**API = Application Programming Interface**

An API allows two software systems to communicate with each other.

Think of an API like a waiter in a restaurant:

```text
You
 ↓
API Request
 ↓
Server
 ↓
API Response
 ↓
You
```

You don't need to know how the server works internally. You only need to know how to send a valid request and understand the response.

### Real-world examples

- Weather app → weather API
- Payment app → payment API
- Maps application → maps API
- Chatbot → AI API
- Your QueryDoc project → LLM/embedding/vector database APIs

## 2. Client and Server

### Client

The client is the application making the request.

Examples:
- Browser
- Mobile app
- Python program
- React application

### Server

The server receives the request, performs some work, and sends a response.

Example:

```text
Python Program (Client)
        |
        | HTTP Request
        v
   API Server
        |
        | JSON Response
        v
Python Program
```

## 3. HTTP and HTTPS

**HTTP** stands for HyperText Transfer Protocol.

**HTTPS** is HTTP with encryption using TLS.

You will commonly see:

```text
http://example.com
https://example.com
```

For real applications, APIs should normally be accessed through **HTTPS** so data is protected during transmission.

You do not need to memorize the networking details yet. Focus on understanding requests and responses.

## 4. URL and Endpoint

A URL identifies where a request is sent.

Example:

```text
https://api.example.com/users
```

An **endpoint** is a specific API URL used for a particular operation.

Examples:

```text
GET /users
GET /users/123
POST /users
```

Think of endpoints as different doors into the API.

## 5. HTTP Methods

The most important HTTP methods are:

| Method | Common purpose |
|---|---|
| GET | Read/fetch data |
| POST | Create/send data |
| PUT | Replace/update a resource |
| PATCH | Partially update a resource |
| DELETE | Delete a resource |

Example:

```text
GET    /users       → get users
POST   /users       → create a user
GET    /users/10    → get user 10
PATCH  /users/10    → update part of user 10
DELETE /users/10    → delete user 10
```

The exact behavior depends on the API documentation.

## 6. Request and Response

An API interaction normally contains:

### Request
- HTTP method
- URL
- Headers
- Query parameters
- Path parameters
- Optional body

### Response
- Status code
- Headers
- Response body

Example:

```text
CLIENT
  |
  | GET /users/10
  |
  v
SERVER
  |
  | 200 OK
  | {"id": 10, "name": "Alex"}
  v
CLIENT
```

## 7. JSON

**JSON = JavaScript Object Notation**

JSON is one of the most common formats used by APIs.

Example:

```json
{
  "name": "Divya",
  "age": 20,
  "skills": ["Python", "Flutter"]
}
```

JSON uses:

- Objects → `{ }`
- Arrays → `[ ]`
- Key-value pairs → `"key": value`
- Strings → quotes
- Boolean values → `true` / `false`
- Empty value → `null`

In Python, JSON objects are commonly represented as dictionaries and JSON arrays as lists.

## 8. Query Parameters

Query parameters provide additional information to an endpoint.

Example:

```text
https://api.example.com/products?category=books&limit=10
```

Here:

```text
category = books
limit = 10
```

The first parameter starts with `?`.

Additional parameters are separated using `&`.

In Python requests, pass them using the `params` argument instead of manually building the URL.

````python
params = {
    "category": "books",
    "limit": 10
}

print(params)
````

## 9. Path Parameters

A path parameter identifies a specific resource.

Example:

```text
/users/123
```

Here `123` can represent a user ID.

Compare:

```text
/users?limit=10
```

Query parameter → filters/configures a request.

```text
/users/123
```

Path parameter → identifies a particular resource.

## 10. Headers

Headers provide additional information about a request or response.

Common request headers:

```text
Content-Type: application/json
Authorization: Bearer <token>
Accept: application/json
```

### Content-Type

Tells the server what format the request body uses.

### Authorization

Used to provide authentication credentials.

### Accept

Tells the server what response format the client prefers.

Never put secret API keys directly into code that you plan to publish.

## 11. HTTP Status Codes

Status codes tell you what happened to the request.

### 2xx — Success

- `200 OK` → successful request
- `201 Created` → resource created
- `204 No Content` → successful response with no response body

### 3xx — Redirection

- `301`
- `302`

### 4xx — Client-side problem

- `400 Bad Request`
- `401 Unauthorized`
- `403 Forbidden`
- `404 Not Found`
- `429 Too Many Requests`

### 5xx — Server-side problem

- `500 Internal Server Error`
- `502 Bad Gateway`
- `503 Service Unavailable`

Easy memory:

```text
2xx → Success
3xx → Redirect
4xx → Your request/client problem
5xx → Server problem
```

## 13. Your First GET Request

For learning, we can use JSONPlaceholder, a public fake REST API designed for testing.

Example endpoint:

```text
https://jsonplaceholder.typicode.com/posts/1
```

Let's request it with Python.

````python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.get(url)

print("Status:", response.status_code)
print("Response:", response.text)
````

## 14. Understanding `response`

The response object contains useful information.

Common attributes:

```python
response.status_code
response.text
response.json()
response.headers
response.url
```

### `response.text`

Returns the response body as text.

### `response.json()`

Parses a JSON response into Python data.

````python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"
response = requests.get(url)

print("Status:", response.status_code)

data = response.json()

print("Python object:", data)
print("Title:", data["title"])
print("User ID:", data["userId"])
````

## 15. GET Request with Query Parameters

Instead of manually writing:

```text
...?userId=1
```

use `params`:

```python
params = {"userId": 1}
requests.get(url, params=params)
```

This is cleaner and safer.

````python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 1
}

response = requests.get(url, params=params)

print("Final URL:", response.url)
print("Status:", response.status_code)

posts = response.json()

print("Number of posts:", len(posts))
print("First post:", posts[0])
````

## 16. POST Request

`POST` is commonly used to send data to a server or create a resource.

JSONPlaceholder accepts POST requests for testing.

The `json=` argument automatically serializes a Python dictionary as JSON and sends the appropriate content type.

````python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

payload = {
    "title": "Learning APIs",
    "body": "I am learning Python API development.",
    "userId": 1
}

response = requests.post(url, json=payload)

print("Status:", response.status_code)
print("Created resource:", response.json())
````

## 17. PUT and PATCH

### PUT

Usually used to replace/update an entire resource.

```python
requests.put(url, json=data)
```

### PATCH

Usually used to update only part of a resource.

```python
requests.patch(url, json=data)
```

The exact behavior depends on the API's documentation.

````python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

patch_data = {
    "title": "Updated title"
}

response = requests.patch(url, json=patch_data)

print("Status:", response.status_code)
print(response.json())
````

## 18. DELETE Request

`DELETE` is commonly used to remove a resource.

Example:

````python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.delete(url)

print("Status:", response.status_code)
````

## 19. Headers with Python

You can send headers using the `headers` argument.

Example:

```python
headers = {
    "Accept": "application/json"
}

response = requests.get(url, headers=headers)
```

For authenticated APIs, the header may look like:

```text
Authorization: Bearer YOUR_TOKEN
```

Do not publish real tokens or API keys.

````python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

headers = {
    "Accept": "application/json"
}

response = requests.get(url, headers=headers)

print(response.status_code)
print(response.json())
````

## 21. Environment Variables

Example `.env`:


For local projects, the `.env` file should normally be added to `.gitignore`.

## 23. API Error Handling

Never assume every request succeeds.

A simple pattern:

```python
response = requests.get(url)

if response.status_code == 200:
    data = response.json()
else:
    print("Request failed")
```

`raise_for_status()` can raise an exception for HTTP error status codes.

````python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()

    data = response.json()
    print("Success:", data)

except requests.exceptions.Timeout:
    print("The request timed out.")

except requests.exceptions.RequestException as error:
    print("Request failed:", error)
````

## 24. Timeout

Always consider using a timeout when making network requests.

Without a timeout, a program may wait longer than expected for a response.

Example:

```python
requests.get(url, timeout=10)
```

This means the request will not wait indefinitely.

## 27. Rate Limits

APIs may limit how many requests you can make within a period.

For example, an API might allow a certain number of requests per minute.

If you receive:

```text
429 Too Many Requests
```

you may have exceeded the API's rate limit.

Production applications should handle rate limiting appropriately, often using retries with backoff when the API documentation allows it.
